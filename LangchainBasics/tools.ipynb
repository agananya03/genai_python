```python
## @tool decorator

from langchain_core.tools import tool

@tool 
def division(a:int, b:int)->int:
    """Divide 2 Number"""
    return a/b
```

```python
print(division.name)
print(division.description)
print(division.args)
```

```python
## async implementation

from langchain_core.tools import tool

@tool
async def amulitply(a:int, b:int)->int:
    """Multiply 2 numbers"""
    return a*b
```

```python
from typing import Annotated, List

@tool
def multiply_by_max(
    a: Annotated[int, "A value"],
    b: Annotated[List[int], "B list value from which to take maximum"]
)->int:
    """Multiply a by maximum of b"""
    return a* max(b)
```

```python
### Structured Tool

from langchain_core.tools import StructuredTool

def multiply(a:int, b:int)->int:
    """Multiply 2 numbers"""
    return a*b 

async def amultiply(a:int, b:int)->int:
    """Multiply 2 numbers."""
    return a*b 

calculator=StructuredTool.from_function(func=multiply, coroutine=amultiply)

print(calculator.invoke({"a":5, "b":6}))
print(await calculator.ainvoke({"a":2, "b":5}))
```

```python
import arxiv
from langchain_core.tools import tool

@tool
def search_arxiv(paper_id: str) -> str:
    """Search for an arXiv paper using its ID."""
    
    client = arxiv.Client()
    
    search = arxiv.Search(
        id_list=[paper_id]
    )
    
    paper = next(client.results(search))
    
    return f"""
Title: {paper.title}

Authors: {', '.join(str(author) for author in paper.authors)}

Published: {paper.published}

Summary: {paper.summary}
"""
```

```python
result = search_arxiv.invoke("1706.03762")

print(result)
```
