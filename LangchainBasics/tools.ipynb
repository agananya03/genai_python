```python
## @tool decorator

from langchain_core.tools import tool

@tool 
def division(a:int, b:int)->int:
    """Divide 2 Number"""
    return a/b
```

```python
print(division.name)
print(division.description)
print(division.args)
```

```python
## async implementation

from langchain_core.tools import tool

@tool
async def amulitply(a:int, b:int)->int:
    """Multiply 2 numbers"""
    return a*b
```

```python
from typing import Annotated, List

@tool
def multiply_by_max(
    a: Annotated[int, "A value"],
    b: Annotated[List[int], "B list value from which to take maximum"]
)->int:
    """Multiply a by maximum of b"""
    return a* max(b)
```

```python
### Structured Tool

from langchain_core.tools import StructuredTool

def multiply(a:int, b:int)->int:
    """Multiply 2 numbers"""
    return a*b 

async def amultiply(a:int, b:int)->int:
    """Multiply 2 numbers."""
    return a*b 

calculator=StructuredTool.from_function(func=multiply, coroutine=amultiply)

print(calculator.invoke({"a":5, "b":6}))
print(await calculator.ainvoke({"a":2, "b":5}))
```
