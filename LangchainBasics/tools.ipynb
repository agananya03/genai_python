```python
## async implementation

from langchain_core.tools import tool

@tool
async def amulitply(a:int, b:int)->int:
    """Multiply 2 numbers"""
    return a*b
```

```python
from typing import Annotated, List

@tool
def multiply_by_max(
    a: Annotated[int, "A value"],
    b: Annotated[List[int], "B list value from which to take maximum"]
)->int:
    """Multiply a by maximum of b"""
    return a* max(b)
```

```python
### Structured Tool

from langchain_core.tools import StructuredTool

def multiply(a:int, b:int)->int:
    """Multiply 2 numbers"""
    return a*b 

async def amultiply(a:int, b:int)->int:
    """Multiply 2 numbers."""
    return a*b 

calculator=StructuredTool.from_function(func=multiply, coroutine=amultiply)

print(calculator.invoke({"a":5, "b":6}))
print(await calculator.ainvoke({"a":2, "b":5}))
```

```python
import arxiv
from langchain_core.tools import tool

@tool
def search_arxiv(paper_id: str) -> str:
    """Search for an arXiv paper using its ID."""
    
    client = arxiv.Client()
    
    search = arxiv.Search(
        id_list=[paper_id]
    )
    
    paper = next(client.results(search))
    
    return f"""
Title: {paper.title}

Authors: {', '.join(str(author) for author in paper.authors)}

Published: {paper.published}

Summary: {paper.summary}
"""
```

```python
result = search_arxiv.invoke("1706.03762")

print(result)
```

### Call Tools with LLM Model

```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

api_wrapper=WikipediaAPIWrapper(top_k_results=5, doc_content_chars_max=500)
wiki_tool=WikipediaQueryRun(api_wrapper=api_wrapper)

print(wiki_tool.invoke({"query": "Langchain"}))
```

```python
from langchain_core.tools import tool

@tool
def add(a:int, b:int)->int:
    """Add a and b"""
    return a+b

@tool
def multiply(a:int, b:int)->int:
    """Multiply a and b"""
    return a*b
```

```python
from langchain_tavily import TavilySearch

tavily_tool=TavilySearch(
    max_results=5,
    topic="general",
    
)
```

```python
tools=[wiki_tool, add, multiply, tavily_tool]
```

```python
from langchain.chat_models import init_chat_model
llm = init_chat_model("qwen/qwen3.6-27b", model_provider="groq")
llm
```

```python
## bind multiple tools with llm
llm_with_tool = llm.bind_tools(tools)
llm_with_tool
```

```python
from langchain_core.messages import HumanMessage, ToolMessage

query="What is 5*6"
messages= [HumanMessage(query)]
response=llm_with_tool.invoke(query)
messages.append(response)
print(response)
```

```python
response.tool_calls
```

```python

for tool_call in response.tool_calls:
    selected_tool = {"add": add, "multiply": multiply, "wikipedia": wiki_tool, "tavily_search": tavily_tool}[tool_call["name"].lower()]
    tool_msg = selected_tool.invoke(tool_call)
    messages.append(tool_msg)

messages
```

```python
messages
```

```python
llm_with_tool.invoke(messages)
```

```python
from langchain_core.messages import HumanMessage
query = "What is Langchain and what is 2*6?"

messages=[HumanMessage(query)]
while True:
    ai_msg = llm_with_tool.invoke(messages)
    messages.append(ai_msg)
    if not ai_msg.tool_calls:
        break
    
    for tool_call in ai_msg.tool_calls:
        selected_tool = {"add": add, "multiply": multiply, "wikipedia": wiki_tool, "tavily_search": tavily_tool}[tool_call["name"].lower()]
        tool_msg = selected_tool.invoke(tool_call)
        messages.append(tool_msg)

```
